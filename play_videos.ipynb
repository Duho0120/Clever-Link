# 영상 순차 재생

아래 `VIDEO_FOLDER` 변수에 영상 파일들이 있는 폴더 경로를 직접 입력한 뒤 순서대로 셀을 실행하세요.

```python
import glob
import os
import functools
import http.server
import socketserver
import threading
from urllib.parse import quote
from IPython.display import display, HTML

# 여기에 영상 파일들이 있는 폴더 경로를 직접 입력하세요
VIDEO_FOLDER = r"Z:\pipeline\AIworkflow-VLM-report\extracted_events\monocular-live\inst-06\FALLEN"

VIDEO_EXTENSIONS = (".mp4", ".mkv", ".mov", ".avi", ".webm", ".wmv")

# 이미 떠 있는 서버가 있으면 먼저 종료 (경로 바꿔서 셀 재실행할 때 포트 충돌 방지)
if "httpd" in globals():
    httpd.shutdown()
    httpd.server_close()

PORT = 8901
_handler = functools.partial(http.server.SimpleHTTPRequestHandler, directory=VIDEO_FOLDER)
httpd = socketserver.TCPServer(("127.0.0.1", PORT), _handler)
threading.Thread(target=httpd.serve_forever, daemon=True).start()
print(f"로컬 서버 시작: http://127.0.0.1:{PORT} (폴더: {VIDEO_FOLDER})")
```

```python
video_files = sorted(
    f for f in glob.glob(os.path.join(VIDEO_FOLDER, "*"))
    if f.lower().endswith(VIDEO_EXTENSIONS)
)

print(f"{len(video_files)}개 영상 발견")
for f in video_files:
    print(os.path.basename(f))
```

```python
# 한 개씩 확인하고 싶을 때: 아래 index만 바꿔가며 이 셀을 재실행
index = 0

name = os.path.basename(video_files[index])
print(name)
url = f"http://127.0.0.1:{PORT}/{quote(name)}"
display(HTML(f'<video controls width="640" src="{url}"></video>'))
```

```python
COLS = 5

items_html = []
for i, f in enumerate(video_files):
    name = os.path.basename(f)
    url = f"http://127.0.0.1:{PORT}/{quote(name)}"
    vid_id = f"vid_{i}"
    items_html.append(f'''
    <div style="flex: 0 0 calc({100/COLS}% - 8px); box-sizing: border-box;">
        <b>{name}</b><br>
        <video id="{vid_id}" controls preload="none" width="100%" src="{url}"></video><br>
        <button onclick="
            const v = document.getElementById('{vid_id}');
            v.classList.toggle('pseudo-fullscreen');
            if (v.classList.contains('pseudo-fullscreen')) {{
                v.style.position = 'fixed';
                v.style.top = '0'; v.style.left = '0';
                v.style.width = '100vw'; v.style.height = '100vh';
                v.style.zIndex = '99999'; v.style.background = 'black';
            }} else {{
                v.style.position = '';
                v.style.top = ''; v.style.left = '';
                v.style.width = '100%'; v.style.height = '';
                v.style.zIndex = ''; v.style.background = '';
            }}
        ">전체화면</button>
    </div>
    ''')

grid_html = (
    '<div style="display: flex; flex-wrap: wrap; gap: 8px;">'
    + "".join(items_html)
    + '</div>'
)

display(HTML(grid_html))
```
